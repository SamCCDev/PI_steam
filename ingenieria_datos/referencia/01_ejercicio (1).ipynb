# CONTAR CANTIDAD DE REGISTROS Y PÓLIZAS ÚNICAS
Necesito es con RDD determinar la cantidad de registros y la cantidad de pólizas únicas

```python
from pyspark import SparkContext
```

```python
sc = SparkContext(master="local", appName="importacionesEnero2017")
```

```python
sc
```

```python
importaciones_ene_2017_RDD = sc.textFile("../data/ene2017_anonimizado.csv")
```

```python
importaciones_ene_2017_RDD.take(5)
```

```python
importaciones_ene_2017_RDD = importaciones_ene_2017_RDD.map(lambda line: line.split("|"))
```
